# Http
http is a package that collects several modules for working with the HyperText Transfer Protocol:

HTTP has 5 methods \
GET POST PUT PATCH DELETE

http.client is a low-level HTTP protocol client; for high-level URL opening use urllib.request [We will use requests for this part]

http.server contains basic HTTP server classes based on socketserver

http.cookies has utilities for implementing state management with cookies

http.cookiejar provides persistence of cookies

# HTTPX

```python
import httpx

response = httpx.get("https://www.example.com")

print(response.status_code)
print(response.headers)
print(response.headers['content-type'])
print(response.text)
```

```python
r = httpx.get('https://httpbin.org/get')
r.text
```

Synchronus Post

```python
r = httpx.post("https://httpbin.org/post" , data={"name": "John", "age": 30})
r.json()
```

Async Post and get

```python
import asyncio

async def main():
    async with httpx.AsyncClient() as client:
        response = await client.get("http://www.example.com")
        print(response.text)

asyncio.run(main())
```

# Requests

```python
from http import HTTPStatus
from http.client import HTTPConnection

status = list(HTTPStatus)
for s in status:
    print(f"{s.value:<6} ->  {s.name:<30}  ->   {s.phrase:<40}")

print(status)

conn = HTTPConnection("www.example.com")
conn.request("GET", "/")
response = conn.getresponse()
response.status

print(response.read())
```

```python
import requests

url = "https://jsonplaceholder.typicode.com/posts/1"

response = requests.get(url)

print(f"Status Code : {response.status_code}")

data = response.json()
print(f"Title : {data['title']}")
print(f"Content : {data['body']}")
```

```python
import json

url = "https://jsonplaceholder.typicode.com/posts"

response = requests.get(url)

print(f"Status Code : {response.status_code}")

# Save data to json file
with open('data.json' , 'w') as file:
    data = response.json()
    json.dump(data , file , indent=2)
```

```python
url = "https://jsonplaceholder.typicode.com/posts"

params = {
    "userId": 1,
    "_limit" :3
}
response = requests.get(url , params = params)

print(f"Status Code : {response.status_code}")

data = response.json()

for post in data:
    print(f"Post {post['id']} -> {post['title']}")
```

```python
new_post = {
    "title": "This is a new post",
    "body" : "This is body of the new post",
    "userId": 1
}

response = requests.post(url , data = new_post)

print(f"Status Code : {response.status_code}")

created_data = response.json()
print(f"Post created by id : {created_data['id']} title : {created_data['title']} and content is {created_data['body']}" )
```

```python
url_patch ="https://jsonplaceholder.typicode.com/posts/1"

new_data = {
    "title" : "This is the new tittle",
    "body" : "This is the new body",
    "userId" : 1
}

response = requests.put(url_patch , json = new_data)
print(f"Status Code : {response.status_code}")

url_patch = "https://jsonplaceholder.typicode.com/posts/1"
partial_update = {
    "title": "Only Updating the Title"
}
response = requests.patch(url_patch, json=partial_update)
print(f"PATCH Status: {response.status_code}")
```

```python
url = "https://jsonplaceholder.typicode.com/posts/1"

response= requests.delete(url)
print(f"Status Code : {response.status_code}")
```
