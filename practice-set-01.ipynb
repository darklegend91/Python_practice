## 1. ATM Transaction Validator

Problem Statement
An ATM processes N withdrawal requests sequentially.
Each request has an amount. Rules:
- Withdrawal amount must be a multiple of 100
- Account balance must never go negative
- For each transaction, print SUCCESS or FAILED

Input

InitialBalance
N
amount1
amount2
...
amountN

Output
SUCCESS
FAILED
SUCCESS
...
FinalBalance

Sample Input:
5000

4
1200
155
2000
2500

Sample Output:
SUCCESS
FAILED
SUCCESS
FAILED
1800

Hint:
Update balance only if both conditions are satisfied.

```python
class InvalidWithdrawalAmount(Exception):
    """Raised when a withdrawal amount is invalid."""


class InvalidAccountBalance(Exception):
    """Raised when an account balance would become negative."""


def check_balance(amount: int) -> bool:
    if amount < 0:
        raise InvalidAccountBalance(
            "Account balance must never go negative."
        )
    return True


def check_withdrawal_amount(amount: int) -> bool:
    if amount <= 0:
        raise InvalidWithdrawalAmount(
            "Withdrawal amount must be greater than zero."
        )

    if amount % 100 != 0:
        raise InvalidWithdrawalAmount(
            "Withdrawal amount must be a multiple of 100."
        )

    return True


try:
    balance = int(input("Enter initial balance: "))
    check_balance(balance)

    transactions = int(input("Enter the number of transactions: "))

    if transactions < 0:
        raise ValueError("Number of transactions cannot be negative.")

    for transaction_number in range(1, transactions + 1):
        try:
            withdrawal_amount = int(
                input(
                    f"Enter amount for transaction "
                    f"{transaction_number}: "
                )
            )

            check_withdrawal_amount(withdrawal_amount)
            check_balance(balance - withdrawal_amount)

            balance -= withdrawal_amount
            print(f"Withdrawal successful. Current balance: {balance}")

        except InvalidWithdrawalAmount as error:
            print(f"Withdrawal failed: {error}")

        except InvalidAccountBalance as error:
            print(f"Withdrawal failed: {error}")

        except ValueError:
            print("Withdrawal failed: Please enter a valid integer.")

    print(f"Final balance: {balance}")

except InvalidAccountBalance as error:
    print(f"Invalid initial balance: {error}")

except ValueError as error:
    print(f"Invalid input: {error}")
```

## 2. Smart Electricity Billing
Problem Statement
Electricity bill is calculated slab-wise:
- First 100 units → ₹3/unit
-  Next 100 units → ₹5/unit
- Remaining → ₹8/unit

If usage > 300 units, add 10% surcharge.
Input:
units
Output:
total_bill

Sample Input
350:

Sample Output:
1540

Hint:
Apply slabs incrementally, then apply surcharge conditionally.

```python
units = int(input("Enter Number of Units: "))

#Calculate the bill
billAmount = 0

if (units <= 100):
    billAmount += units * 3
    
elif (100 < units <=200):
    billAmount += 300 + ((units-100) * 5) 

elif (200 < units <=300):
    billAmount += 800 + ((units - 200) * 8)
    
else:
    billAmount += 1600 + ((units - 300) * 8)
    billAmount += ( billAmount* 0.1)
    
print(billAmount)
```

## 5️. Salary Deduction System
Problem Statement\
Employee salary rules:
- Basic salary given
- If late days > 5 → deduct 5%
- If late days > 10 → deduct 10%
- If absent days > 2 → deduct additional 5%

Input:\
salary\
late_days\
absent_days

Output:\
final_salary

Sample Input:
50000\
8\
1

Sample Output:\
47500

Hint:\
Apply deductions cumulatively, not exclusively.

```python
salary = int(input("Enter the Base Salary: "))
late_days = int(input("Enter the late_days: "))
absent_days = int(input("Enter the absent_days: "))
deduction_perc = 0

if (late_days > 5):
    deduction_perc = 0.05

if (late_days > 10):
    deduction_perc += 0.1

if (absent_days  > 2):
    deduction_perc += 0.1

final_salary = salary - salary*(deduction_perc)
print(final_salary)
```

## 6️. Prime Range Analyzer
Problem Statement
Print count of prime numbers between A and B (inclusive).
Input:
A
B

Output:\
prime_count

Sample Input:\
10\
30

Sample Output:\
6

Hint:\
Check divisibility up to √n.

```python
import math as math

def isPrime(num : int) -> bool:
    for j in range (2 , math.isqrt(num) + 1):
        if (num % j == 0 ):
            return False
    return True

start_num = int(input("Enter the starting number: "))
end_num = int(input("Enter the ending number: "))

count = 0

for i in range(start_num , end_num+1):
    if (isPrime(i)):
        count +=1
    
print(f"The count of prime numbers form {start_num} to {end_num} is {count}.")
```

## 17. Armstrong Number Checker

Check if a number equals sum of cubes of its digits.

Input\
number

Output\
YES / NO

Sample Input\
153

Sample Output\
YES

Hint:\
Extract digits using modulo and division.

```python
#Define the isArmstrong function
def isArmstrong(number : int) ->bool:
    original_num = number
    armstrong_num=0
    
    while(number > 0):
        armstrong_num += (number % 10) ** 3
        number =int( number / 10)
    
    
    if (original_num == armstrong_num):
        return True
    else:
        return False

num = int(input("Enter the number: "))
if isArmstrong(num):
    print("YES")
else:
    print("NO")
```

## 19. Number Mirror Validator

Reverse a number and check if original equals reversed.\

Input\
number

Output\
PALINDROME / NOT PALINDROME

Sample Input\
1221

Sample Output\
PALINDROME

Hint:\
Build reverse using arithmetic.

```python

def reversed_num(num:int) -> int:
    r_num = 0
    while(num > 0):
        r_num = (r_num*10) + ( num %10)
        num = int(num / 10)
    return r_num

num = int(input("Enter the number: "))

if (reversed_num(num) == num):
    print("PALINDROME")

else:
    print("NOT PALINDROME")
```

## 20. Digital Lock Countdown

Given a number, repeatedly subtract sum of its digits until result is a single digit.

Input\
number

Output\
final_digit

Sample Input\
987

Sample Output\
6

Hint:\
Nested loops: digit sum inside reduction loop.

```python
number = int(input())

while (number >= 10):
    digit_sum = 0

    while (number > 0):
        digit_sum += number % 10
        number = number //10

    number = digit_sum

print(number)
```
